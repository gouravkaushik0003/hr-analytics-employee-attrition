```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# CSV file load karein
df = pd.read_csv('WA_Fn-UseC_-HR-Employee-Attrition.csv')

# Top 5 rows dekhein
df.head()
```

```python
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x='Department', hue='Attrition', palette='Set2')
plt.title('Department-wise Employee Attrition')
plt.xlabel('Department')
plt.ylabel('Employee Count')
plt.show()
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='OverTime', hue='Attrition', palette='coolwarm')
plt.title('Impact of OverTime on Attrition')
plt.xlabel('OverTime')
plt.ylabel('Employee Count')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sirf numerical columns select karein correlation ke liye
plt.figure(figsize=(10, 8))
numeric_df = df.select_dtypes(include=['int64', 'float64'])
sns.heatmap(numeric_df.corr(), annot=False, cmap='coolwarm', linewidths=0.5)
plt.title('Feature Correlation Heatmap')
plt.show()
```
